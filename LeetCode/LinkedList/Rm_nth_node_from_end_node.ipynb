```python
from typing import Optional

class Solution :
    def removeNthFromEnd(self, head: Optional[Node], n: int) -> Optional[Node] :
        # the size of the likedlist
        current = head
        size = 1
        while current.next != None :
            current = current.next
            size += 1
        if size < n :
            return -1
        # the positon of the nth node
        target = size - n

        # reach out to the target
        curr = head
        for _ in range(target) :
            curr = curr.next
        # remove the nth node from last
        curr.next = curr.next.next

```
