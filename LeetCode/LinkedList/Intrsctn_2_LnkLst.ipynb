```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

    def __len__(self) :
        count = 0
        current = self
        while current :
            count += 1
            current = current.next
        return count
```

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def getIntersectionNode(self, headA: ListNode, headB: ListNode) -> Optional[ListNode]:
        
        while headA.next != headB.next :
            headA = headA.next
            headB = headB.next
        return headA.val
        
```
