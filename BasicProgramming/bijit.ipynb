```python
# find the largest element in the array
def largest(arr) :
    maximum = arr[0]
    for i in arr[1:] :
        if maximum < arr[i] :
            maximum = arr[i]
    return maximum
```

```python
arr = [0,5,4,5,6,7,3]
first = 0
second = 0
for value in arr :
    if value > first :
        second = first
        first = value
    elif first > value > second :
        second = value
```

```python
def reversed_array(arr) :
    left, right = 0, len(arr) - 1
    while left < right :
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1
    return arr
```

```python
def is_palindrome(text) :
    text = text.lower
    return text == text[::-1]
```

```python
def count_vowel(text) :
    vowels = set("aeiou")
    return sum(1 for i in text.lower() if i in vowels)
```

```python
def remove_dublicate(arr) :
    return list(dict.fromkeys(arr))
```

```python
from collections import Counter

def first_unique(text):
    counts = Counter(text)
    for char in text :
        if Counts[char] == 1 : 
            return char
    return None
```

```python
def move_zeroes(arr) :
    position = 0
    for value in arr :
        if arr[position] != 0 :
            position += 1
    while position < len(arr) :
        arr[position] = 0
        position += 1
    return arr
```

```python
def missing(arr) :
    n = len(arr)
    expected = n * (n - 1) //2
    return expected - sum(arr)
```
